```python
sales.fillna(sales.median(numeric_only=True), inplace=True)
sales.dropna(inplace= True)
sales.shape
```

```python
#sales.drop(*['order_date', 'last_order_date'],axis=1, inplace= True)
sales.drop('order_date',axis=1, inplace= True)
sales.drop('customer_id',axis=1, inplace= True)
sales.drop('Unnamed: 0',axis=1, inplace= True)
```

```python
# ============================================================
# COMPLETE MULTICLASS LIGHTGBM + SHAP ANALYSIS
# ============================================================
#
# Target:
#     product_id
#
# Approximately:
#     500 product classes
#
# Dataset:
#     sales
#
# Split:
#     80% TRAIN
#     20% TEST
#
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import os
import warnings

warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import shap

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    top_k_accuracy_score
)

from lightgbm import LGBMClassifier


# ============================================================
# 1. SETTINGS
# ============================================================

TARGET = "product_id"

TEST_SIZE = 0.5

RANDOM_STATE = 42

TOP_GLOBAL_FEATURES = 30

TOP_PRODUCTS_TO_ANALYZE = 10

TOP_FEATURES_PER_PRODUCT = 20

# For large datasets, calculate SHAP on a sample
# instead of the entire test dataset.
#
# Set to None if you want all test rows.
SHAP_SAMPLE_SIZE = 20000


# ============================================================
# 2. CREATE OUTPUT DIRECTORY
# ============================================================

OUTPUT_DIR = "multiclass_shap_results"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

print("=" * 80)
print("MULTICLASS LIGHTGBM + SHAP ANALYSIS")
print("=" * 80)


# ============================================================
# 3. COPY DATA
# ============================================================

df = sales.copy()

print("\nOriginal dataset shape:")
print(df.shape)


# ============================================================
# 4. BASIC VALIDATION
# ============================================================

if TARGET not in df.columns:

    raise ValueError(
        f"Target column '{TARGET}' not found in dataframe."
    )


# Remove rows where target is missing

missing_target = df[TARGET].isna().sum()

print(
    "\nMissing target rows:",
    missing_target
)


if missing_target > 0:

    df = df.dropna(
        subset=[TARGET]
    ).copy()


# ============================================================
# 5. TARGET DISTRIBUTION
# ============================================================

class_counts = (
    df[TARGET]
    .value_counts()
    .sort_values()
)


print("\n" + "=" * 80)
print("TARGET INFORMATION")
print("=" * 80)

print(
    "\nNumber of unique products:",
    df[TARGET].nunique()
)

print(
    "\nProduct frequency statistics:"
)

print(
    class_counts.describe()
)


print(
    "\nProducts with exactly 1 observation:",
    (class_counts == 1).sum()
)

print(
    "Products with < 5 observations:",
    (class_counts < 5).sum()
)

print(
    "Products with < 10 observations:",
    (class_counts < 10).sum()
)


# ============================================================
# 6. REMOVE ONLY SINGLE-OBSERVATION PRODUCTS
# ============================================================
#
# Stratified 80/20 split requires at least 2 rows
# per class.
#
# We remove ONLY classes with one observation.
#
# ============================================================

valid_classes = class_counts[
    class_counts >= 2
].index


df = df[
    df[TARGET].isin(valid_classes)
].copy()


print("\nAfter removing single-observation products:")

print(
    "Rows:",
    len(df)
)

print(
    "Products:",
    df[TARGET].nunique()
)


# ============================================================
# 7. SEPARATE X AND Y
# ============================================================

X = df.drop(
    columns=[TARGET]
).copy()

y = df[TARGET].copy()


# ============================================================
# 8. REMOVE OBVIOUS ID COLUMNS
# ============================================================
#
# DO NOT automatically remove customer_id/order_id if
# they are intentionally useful.
#
# product_id is already removed because it is TARGET.
#
# ============================================================

print("\nFeatures before preprocessing:")

print(
    X.shape[1]
)


# If these identifiers should NOT be used as predictors,
# uncomment the following:
#
# ID_COLUMNS = [
#     "order_id",
#     "customer_id"
# ]
#
# X = X.drop(
#     columns=[
#         c for c in ID_COLUMNS
#         if c in X.columns
#     ]
# )


# ============================================================
# 9. HANDLE DATETIME FEATURES
# ============================================================

datetime_cols = X.select_dtypes(
    include=[
        "datetime64[ns]",
        "datetime64[ns, UTC]"
    ]
).columns.tolist()


print("\nDatetime columns:")

print(
    datetime_cols
)


for col in datetime_cols:

    X[col + "_year"] = X[col].dt.year

    X[col + "_month"] = X[col].dt.month

    X[col + "_day"] = X[col].dt.day

    X[col + "_dayofweek"] = X[col].dt.dayofweek

    X[col + "_dayofyear"] = X[col].dt.dayofyear

    X[col + "_week"] = (
        X[col]
        .dt.isocalendar()
        .week
        .astype(float)
    )

    X[col + "_hour"] = X[col].dt.hour

    X[col + "_minute"] = X[col].dt.minute

    X[col + "_is_weekend"] = (
        X[col].dt.dayofweek >= 5
    ).astype(int)

    X.drop(
        columns=[col],
        inplace=True
    )


# ============================================================
# 10. HANDLE OBJECT / CATEGORY COLUMNS
# ============================================================

categorical_cols = X.select_dtypes(
    include=[
        "object",
        "category"
    ]
).columns.tolist()


print("\nCategorical columns:")

print(
    categorical_cols
)


for col in categorical_cols:

    X[col] = X[col].astype("category")


# ============================================================
# 11. HANDLE INFINITE VALUES
# ============================================================

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


# ============================================================
# 12. CHECK NUMERIC COLUMNS
# ============================================================

numeric_cols = X.select_dtypes(
    include=np.number
).columns.tolist()


print(
    "\nNumber of numeric features:",
    len(numeric_cols)
)

print(
    "Number of categorical features:",
    len(categorical_cols)
)

print(
    "Total features:",
    X.shape[1]
)


# ============================================================
# 13. TRAIN / TEST SPLIT
# ============================================================

print("\n" + "=" * 80)
print("80 / 20 TRAIN TEST SPLIT")
print("=" * 80)


X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=TEST_SIZE,

    random_state=RANDOM_STATE,

    stratify=y
)


print(
    "\nX_train:",
    X_train.shape
)

print(
    "X_test :",
    X_test.shape
)

print(
    "y_train:",
    y_train.shape
)

print(
    "y_test :",
    y_test.shape
)


print(
    "\nTrain products:",
    y_train.nunique()
)

print(
    "Test products:",
    y_test.nunique()
)


# ============================================================
# 14. CHECK FOR MISSING CLASSES
# ============================================================

train_classes = set(
    y_train.unique()
)

test_classes = set(
    y_test.unique()
)

missing_from_train = (
    test_classes - train_classes
)


if len(missing_from_train) > 0:

    print(
        "\nWARNING:"
    )

    print(
        "Products present in test but not train:"
    )

    print(
        missing_from_train
    )

else:

    print(
        "\nAll test products are represented in training."
    )


# ============================================================
# 15. CREATE MULTICLASS LIGHTGBM MODEL
# ============================================================

NUM_CLASSES = y.nunique()


print("\n" + "=" * 80)
print("LIGHTGBM MODEL")
print("=" * 80)

print(
    "Number of classes:",
    NUM_CLASSES
)


model = LGBMClassifier(

    objective="multiclass",

    num_class=NUM_CLASSES,

    n_estimators=500,

    learning_rate=0.03,

    num_leaves=31,

    max_depth=-1,

    min_child_samples=20,

    subsample=0.8,

    colsample_bytree=0.8,

    reg_alpha=1.0,

    reg_lambda=1.0,

    random_state=RANDOM_STATE,

    n_jobs=-1,

    verbosity=-1
)


# ============================================================
# 16. TRAIN MODEL
# ============================================================

print("\nTraining model...")

model.fit(

    X_train,

    y_train,

    categorical_feature=categorical_cols
)


print(
    "Model training completed."
)


# ============================================================
# 17. MODEL PREDICTION
# ============================================================

print("\nGenerating predictions...")


y_pred = model.predict(
    X_test
)


y_prob = model.predict_proba(
    X_test
)


print(
    "Probability matrix shape:",
    y_prob.shape
)


# ============================================================
# 18. MODEL CLASSES
# ============================================================

classes = model.classes_


print(
    "\nNumber of model classes:",
    len(classes)
)


# ============================================================
# 19. MODEL PERFORMANCE
# ============================================================

print("\n" + "=" * 80)
print("MODEL PERFORMANCE")
print("=" * 80)


accuracy = accuracy_score(
    y_test,
    y_pred
)


balanced_accuracy = balanced_accuracy_score(
    y_test,
    y_pred
)


print(
    "\nAccuracy:",
    round(accuracy, 4)
)


print(
    "Balanced Accuracy:",
    round(
        balanced_accuracy,
        4
    )
)


# ============================================================
# 20. TOP-K ACCURACY
# ============================================================

try:

    top3 = top_k_accuracy_score(

        y_test,

        y_prob,

        k=3,

        labels=classes
    )

    top5 = top_k_accuracy_score(

        y_test,

        y_prob,

        k=5,

        labels=classes
    )

    top10 = top_k_accuracy_score(

        y_test,

        y_prob,

        k=10,

        labels=classes
    )


    print(
        "\nTop-3 Accuracy:",
        round(top3, 4)
    )

    print(
        "Top-5 Accuracy:",
        round(top5, 4)
    )

    print(
        "Top-10 Accuracy:",
        round(top10, 4)
    )

except Exception as e:

    print(
        "\nTop-K accuracy could not be calculated:",
        e
    )


# ============================================================
# 21. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

report = classification_report(

    y_test,

    y_pred,

    digits=4,

    zero_division=0
)


print(
    report
)


# Save report

with open(

    os.path.join(
        OUTPUT_DIR,
        "classification_report.txt"
    ),

    "w"
) as f:

    f.write(report)


# ============================================================
# 22. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_pred,

    labels=classes
)


print(
    "\nConfusion matrix shape:",
    cm.shape
)


# Save confusion matrix

cm_df = pd.DataFrame(

    cm,

    index=classes,

    columns=classes
)


cm_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "confusion_matrix.csv"
    )
)


# ============================================================
# 23. MULTICLASS ROC AUC
# ============================================================

try:

    roc_auc = roc_auc_score(

        y_test,

        y_prob,

        multi_class="ovr",

        average="weighted",

        labels=classes
    )


    print(
        "\nWeighted ROC-AUC:",
        round(
            roc_auc,
            4
        )
    )

except Exception as e:

    print(
        "\nROC-AUC unavailable:",
        e
    )


# ============================================================
# 24. PREDICTION RESULTS
# ============================================================

prediction_results = X_test.copy()


prediction_results["actual_product_id"] = (
    y_test
)


prediction_results["predicted_product_id"] = (
    y_pred
)


prediction_results["prediction_confidence"] = (
    y_prob.max(axis=1)
)


prediction_results["correct_prediction"] = (

    y_test.values == y_pred

)


prediction_results.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "test_predictions.csv"
    ),

    index=True
)


# ============================================================
# 25. SHAP SAMPLE
# ============================================================
#
# With 500 classes, calculating SHAP for every test row
# can consume a lot of memory.
#
# We therefore sample up to 20,000 test observations.
#
# ============================================================

if (
    SHAP_SAMPLE_SIZE is not None
    and len(X_test) > SHAP_SAMPLE_SIZE
):

    X_shap = X_test.sample(

        n=SHAP_SAMPLE_SIZE,

        random_state=RANDOM_STATE
    )

else:

    X_shap = X_test.copy()


print("\n" + "=" * 80)
print("SHAP ANALYSIS")
print("=" * 80)


print(
    "\nSHAP sample:",
    X_shap.shape
)


# ============================================================
# 26. CREATE SHAP EXPLAINER
# ============================================================

print(
    "\nCreating TreeExplainer..."
)


explainer = shap.TreeExplainer(
    model
)


# ============================================================
# 27. CALCULATE SHAP VALUES
# ============================================================

print(
    "\nCalculating SHAP values..."
)

print(
    "This may take some time for ~500 classes."
)


shap_values = explainer.shap_values(
    X_shap
)


# ============================================================
# 28. STANDARDIZE SHAP FORMAT
# ============================================================
#
# Depending on SHAP version:
#
# Old:
#     list of arrays
#
# New:
#     ndarray
#
# We convert everything to:
#
#     samples × features × classes
#
# ============================================================

if isinstance(
    shap_values,
    list
):

    shap_array = np.stack(

        shap_values,

        axis=2
    )

else:

    shap_array = np.asarray(
        shap_values
    )


print(
    "\nSHAP array shape:",
    shap_array.shape
)


# ============================================================
# 29. VALIDATE SHAP SHAPE
# ============================================================

expected_features = X_shap.shape[1]

expected_samples = X_shap.shape[0]

expected_classes = len(classes)


if shap_array.shape != (

    expected_samples,

    expected_features,

    expected_classes

):

    raise ValueError(

        f"Unexpected SHAP shape: "
        f"{shap_array.shape}. "
        f"Expected: "
        f"({expected_samples}, "
        f"{expected_features}, "
        f"{expected_classes})"
    )


print(
    "SHAP shape validated."
)


# ============================================================
# 30. GLOBAL SHAP IMPORTANCE
# ============================================================
#
# Average absolute SHAP across:
#
#     samples
#     classes
#
# Result:
#
#     one importance score per feature
#
# ============================================================

global_importance = pd.DataFrame({

    "feature": X_shap.columns,

    "mean_abs_shap": np.abs(
        shap_array
    ).mean(
        axis=(0, 2)
    ),

    "max_abs_shap": np.abs(
        shap_array
    ).max(
        axis=(0, 2)
    )

})


global_importance = (

    global_importance

    .sort_values(

        "mean_abs_shap",

        ascending=False
    )

    .reset_index(drop=True)

)


print("\n" + "=" * 80)
print("GLOBAL SHAP IMPORTANCE")
print("=" * 80)


print(
    global_importance.head(
        TOP_GLOBAL_FEATURES
    )
)


global_importance.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "global_shap_importance.csv"
    ),

    index=False
)


# ============================================================
# 31. GLOBAL SHAP BAR PLOT
# ============================================================

plt.figure(
    figsize=(12, 10)
)


shap.summary_plot(

    np.abs(
        shap_array
    ).mean(axis=2),

    X_shap,

    plot_type="bar",

    max_display=TOP_GLOBAL_FEATURES,

    show=False
)


plt.title(
    "Global SHAP Feature Importance - All Products"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "global_shap_bar.png"
    ),

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ============================================================
# 32. GLOBAL BEESWARM
# ============================================================
#
# For a multiclass model, we aggregate class-specific
# SHAP values using mean absolute SHAP.
#
# This gives an overall view of feature importance.
#
# ============================================================

global_abs_shap = np.abs(
    shap_array
).mean(axis=2)


plt.figure(
    figsize=(12, 10)
)


shap.summary_plot(

    global_abs_shap,

    X_shap,

    max_display=TOP_GLOBAL_FEATURES,

    show=False
)


plt.title(
    "Global SHAP Summary - All Product Classes"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "global_shap_summary.png"
    ),

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ============================================================
# 33. LIGHTGBM FEATURE IMPORTANCE
# ============================================================

lgb_importance = pd.DataFrame({

    "feature": X_train.columns,

    "lightgbm_importance":
        model.feature_importances_

})


# ============================================================
# 34. COMPARE LIGHTGBM AND SHAP
# ============================================================

importance_comparison = (

    global_importance

    .merge(

        lgb_importance,

        on="feature",

        how="left"
    )

    .sort_values(

        "mean_abs_shap",

        ascending=False
    )

)


print("\n" + "=" * 80)
print("LIGHTGBM vs SHAP")
print("=" * 80)


print(
    importance_comparison.head(30)
)


importance_comparison.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "lightgbm_vs_shap.csv"
    ),

    index=False
)


# ============================================================
# 35. SELECT PRODUCTS FOR DETAILED SHAP
# ============================================================
#
# We don't want to generate 500 separate plots.
#
# We select the most frequently occurring products
# in the SHAP sample.
#
# ============================================================

products_to_analyze = (

    y.loc[X_shap.index]

    .value_counts()

    .head(
        TOP_PRODUCTS_TO_ANALYZE
    )

    .index

)


print("\n" + "=" * 80)
print("PRODUCTS SELECTED FOR DETAILED SHAP")
print("=" * 80)


print(
    products_to_analyze.tolist()
)


# ============================================================
# 36. CLASS-SPECIFIC SHAP IMPORTANCE
# ============================================================

class_importance_records = []


for class_index, class_value in enumerate(classes):

    class_shap = shap_array[
        :,
        :,
        class_index
    ]


    mean_abs = np.abs(
        class_shap
    ).mean(axis=0)


    mean_signed = class_shap.mean(
        axis=0
    )


    temp = pd.DataFrame({

        "product_id": class_value,

        "feature": X_shap.columns,

        "mean_abs_shap": mean_abs,

        "mean_shap": mean_signed

    })


    temp = temp.sort_values(

        "mean_abs_shap",

        ascending=False
    )


    class_importance_records.append(
        temp
    )


class_importance = pd.concat(

    class_importance_records,

    ignore_index=True
)


class_importance.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "product_class_shap_importance.csv"
    ),

    index=False
)


# ============================================================
# 37. PRINT TOP FEATURES FOR SELECTED PRODUCTS
# ============================================================

print("\n" + "=" * 80)
print("TOP FEATURES PER PRODUCT")
print("=" * 80)


for product in products_to_analyze:

    temp = class_importance[

        class_importance["product_id"]
        == product

    ].head(
        TOP_FEATURES_PER_PRODUCT
    )


    print(
        f"\nProduct ID: {product}"
    )


    print(
        temp[
            [
                "feature",
                "mean_abs_shap",
                "mean_shap"
            ]
        ].to_string(
            index=False
        )
    )


# ============================================================
# 38. CLASS-SPECIFIC SHAP PLOTS
# ============================================================

for product in products_to_analyze:

    class_index_array = np.where(

        classes == product

    )[0]


    if len(class_index_array) == 0:

        continue


    class_index = class_index_array[0]


    class_shap = shap_array[
        :,
        :,
        class_index
    ]


    plt.figure(
        figsize=(12, 10)
    )


    shap.summary_plot(

        class_shap,

        X_shap,

        max_display=TOP_FEATURES_PER_PRODUCT,

        show=False
    )


    plt.title(

        f"SHAP Summary - Product ID {product}"

    )


    plt.tight_layout()


    filename = (

        f"shap_product_{product}.png"

    )


    plt.savefig(

        os.path.join(
            OUTPUT_DIR,
            filename
        ),

        dpi=300,

        bbox_inches="tight"
    )


    plt.show()


# ============================================================
# 39. POSITIVE / NEGATIVE SHAP BY PRODUCT
# ============================================================

impact_records = []


for class_index, product in enumerate(classes):

    class_shap = shap_array[
        :,
        :,
        class_index
    ]


    mean_shap = class_shap.mean(
        axis=0
    )


    mean_abs_shap = np.abs(
        class_shap
    ).mean(
        axis=0
    )


    temp = pd.DataFrame({

        "product_id": product,

        "feature": X_shap.columns,

        "mean_shap": mean_shap,

        "mean_abs_shap": mean_abs_shap

    })


    temp["impact"] = np.where(

        temp["mean_shap"] > 0,

        "Pushes toward this product",

        "Pushes away from this product"

    )


    impact_records.append(
        temp
    )


product_impact = pd.concat(

    impact_records,

    ignore_index=True
)


product_impact.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "product_feature_shap_impact.csv"
    ),

    index=False
)


# ============================================================
# 40. DEPENDENCE PLOTS
# ============================================================
#
# Generate dependence plots for top global features.
#
# We use the strongest product class for each feature.
#
# ============================================================

top_features = (

    global_importance

    .head(10)

    ["feature"]

    .tolist()

)


print("\n" + "=" * 80)
print("DEPENDENCE PLOTS")
print("=" * 80)


for feature in top_features:

    feature_index = X_shap.columns.get_loc(
        feature
    )


    # Find product class where this feature
    # has maximum average absolute SHAP

    class_strength = np.abs(

        shap_array[
            :,
            feature_index,
            :
        ]

    ).mean(
        axis=0
    )


    strongest_class_index = np.argmax(
        class_strength
    )


    strongest_product = classes[
        strongest_class_index
    ]


    print(
        f"\nFeature: {feature}"
    )

    print(
        f"Strongest product class: "
        f"{strongest_product}"
    )


    plt.figure(
        figsize=(10, 7)
    )


    shap.dependence_plot(

        feature,

        shap_array[
            :,
            :,
            strongest_class_index
        ],

        X_shap,

        show=False
    )


    plt.title(

        f"SHAP Dependence - {feature}\n"
        f"Strongest Product Class = "
        f"{strongest_product}"

    )


    plt.tight_layout()


    safe_feature = (

        str(feature)

        .replace("/", "_")

        .replace("\\", "_")

        .replace(" ", "_")

        .replace(":", "_")

    )


    plt.savefig(

        os.path.join(

            OUTPUT_DIR,

            f"dependence_{safe_feature}.png"

        ),

        dpi=300,

        bbox_inches="tight"

    )


    plt.show()


# ============================================================
# 41. LOCAL EXPLANATION
# ============================================================
#
# Explain one individual prediction.
#
# We select the first SHAP observation.
#
# ============================================================

local_position = 0


local_index = X_shap.index[
    local_position
]


X_single = X_shap.iloc[
    [local_position]
]


actual_product = y.loc[
    local_index
]


predicted_product = model.predict(
    X_single
)[0]


prediction_probabilities = model.predict_proba(
    X_single
)[0]


predicted_class_index = np.argmax(
    prediction_probabilities
)


predicted_probability = (
    prediction_probabilities[
        predicted_class_index
    ]
)


print("\n" + "=" * 80)
print("LOCAL PREDICTION EXPLANATION")
print("=" * 80)


print(
    "\nActual product:",
    actual_product
)


print(
    "Predicted product:",
    predicted_product
)


print(
    "Prediction probability:",
    round(
        predicted_probability,
        6
    )
)


# ============================================================
# 42. SHAP FOR PREDICTED PRODUCT
# ============================================================

single_class_shap = shap_array[

    local_position,

    :,

    predicted_class_index

]


# ============================================================
# 43. BASE VALUE
# ============================================================

expected_value = explainer.expected_value


if isinstance(

    expected_value,

    (list, np.ndarray)

):

    base_value = expected_value[
        predicted_class_index
    ]

else:

    base_value = expected_value


# ============================================================
# 44. LOCAL WATERFALL
# ============================================================

single_explanation = shap.Explanation(

    values=single_class_shap,

    base_values=base_value,

    data=X_single.iloc[0],

    feature_names=X_shap.columns

)


plt.figure(
    figsize=(12, 10)
)


shap.plots.waterfall(

    single_explanation,

    max_display=20,

    show=False
)


plt.tight_layout()


plt.savefig(

    os.path.join(

        OUTPUT_DIR,

        "local_waterfall.png"

    ),

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 45. LOCAL FEATURE IMPACT TABLE
# ============================================================

local_explanation = pd.DataFrame({

    "feature": X_shap.columns,

    "feature_value":
        X_single.iloc[0].values,

    "shap_value":
        single_class_shap

})


local_explanation = (

    local_explanation

    .sort_values(

        "shap_value",

        ascending=False

    )

)


print("\n" + "=" * 80)
print("TOP FEATURES PUSHING TOWARD PREDICTED PRODUCT")
print("=" * 80)


print(

    local_explanation.head(20)

)


print("\n" + "=" * 80)
print("TOP FEATURES PUSHING AWAY FROM PREDICTED PRODUCT")
print("=" * 80)


print(

    local_explanation.tail(20)

)


local_explanation.to_csv(

    os.path.join(

        OUTPUT_DIR,

        "local_shap_explanation.csv"

    ),

    index=False

)


# ============================================================
# 46. TOP FEATURES PER PRODUCT - WIDE FORMAT
# ============================================================
#
# Useful for business analysis.
#
# ============================================================

top_product_features = (

    class_importance

    .sort_values(

        [
            "product_id",
            "mean_abs_shap"
        ],

        ascending=[
            True,
            False
        ]

    )

    .groupby(
        "product_id"
    )

    .head(
        TOP_FEATURES_PER_PRODUCT
    )

)


top_product_features.to_csv(

    os.path.join(

        OUTPUT_DIR,

        "top_features_per_product.csv"

    ),

    index=False

)


# ============================================================
# 47. FEATURE RANKING
# ============================================================

global_importance["rank"] = (

    global_importance

    ["mean_abs_shap"]

    .rank(

        ascending=False,

        method="min"

    )

    .astype(int)

)


global_importance.to_csv(

    os.path.join(

        OUTPUT_DIR,

        "final_global_feature_ranking.csv"

    ),

    index=False

)


# ============================================================
# 48. SAVE SHAP VALUES FOR TOP PRODUCTS ONLY
# ============================================================
#
# Do NOT save 500 class × every row unnecessarily.
#
# We save detailed SHAP matrices only for the selected
# important/frequent products.
#
# ============================================================

for product in products_to_analyze:

    class_index_array = np.where(

        classes == product

    )[0]


    if len(class_index_array) == 0:

        continue


    class_index = class_index_array[0]


    product_shap_df = pd.DataFrame(

        shap_array[
            :,
            :,
            class_index
        ],

        columns=X_shap.columns,

        index=X_shap.index

    )


    product_shap_df.to_csv(

        os.path.join(

            OUTPUT_DIR,

            f"shap_values_product_{product}.csv"

        ),

        index=True

    )


# ============================================================
# 49. MODEL SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("FINAL SUMMARY")
print("=" * 80)


print(
    "\nOriginal rows:",
    len(sales)
)


print(
    "Rows used:",
    len(df)
)


print(
    "Products:",
    NUM_CLASSES
)


print(
    "Features:",
    X.shape[1]
)


print(
    "Training rows:",
    len(X_train)
)


print(
    "Testing rows:",
    len(X_test)
)


print(
    "SHAP rows:",
    len(X_shap)
)


print(
    "\nAccuracy:",
    round(
        accuracy,
        4
    )
)


print(
    "\nTop global SHAP features:"
)


print(

    global_importance.head(20)[

        [
            "rank",
            "feature",
            "mean_abs_shap"
        ]

    ].to_string(
        index=False
    )

)


print("\n")
print("=" * 80)
print("FILES SAVED")
print("=" * 80)


print(
    f"\nOutput directory:"
    f" {OUTPUT_DIR}"
)


print(
    """
1. global_shap_importance.csv
2. final_global_feature_ranking.csv
3. product_class_shap_importance.csv
4. product_feature_shap_impact.csv
5. top_features_per_product.csv
6. lightgbm_vs_shap.csv
7. test_predictions.csv
8. confusion_matrix.csv
9. classification_report.txt
10. global_shap_bar.png
11. global_shap_summary.png
12. shap_product_<product>.png
13. dependence_<feature>.png
14. local_waterfall.png
15. local_shap_explanation.csv
16. shap_values_product_<product>.csv
"""
)



print("\nSHAP ANALYSIS COMPLETED.")
```
