# 📊 Customer 20-Feature Extractor from `Order_Details` Database

This notebook connects to the database (`Order_Details` and `Customer_Details` tables), accepts any `customer_id` input, verifies customer existence, and computes **exactly the 20 SHAP features** required for model prediction.

### 🎯 The 20 Required Features:
1. `min_pp`: Minimum Product Price
2. `p90_pp`: 90th Percentile Product Price
3. `p95_pad`: 95th Percentile Price After Discount
4. `p25_pp`: 25th Percentile Product Price
5. `p95_pp`: 95th Percentile Product Price
6. `max_pp`: Maximum Product Price
7. `min_pad`: Minimum Price After Discount
8. `p75_pp`: 75th Percentile Product Price
9. `max_pad`: Maximum Price After Discount
10. `p90_pad`: 90th Percentile Price After Discount
11. `avg_pp`: Average Product Price
12. `p50_pp`: 50th Percentile (Median) Product Price
13. `p25_pad`: 25th Percentile Price After Discount
14. `p50_pad`: 50th Percentile (Median) Price After Discount
15. `avg_pad`: Average Price After Discount
16. `p75_pad`: 75th Percentile Price After Discount
17. `p90_pd`: 90th Percentile Product Discount
18. `total_pad`: Total Sum of Price After Discount
19. `max_pd`: Maximum Product Discount
20. `p95_pd`: 95th Percentile Product Discount

```python
import os
import pickle
import numpy as np
import pandas as pd
import db_manager

# Exact 20 SHAP features in required model input order
SELECTED_20_FEATURES = [
    'min_pp', 'p90_pp', 'p95_pad', 'p25_pp', 'p95_pp', 'max_pp',
    'min_pad', 'p75_pp', 'max_pad', 'p90_pad', 'avg_pp', 'p50_pp',
    'p25_pad', 'p50_pad', 'avg_pad', 'p75_pad', 'p90_pd', 'total_pad',
    'max_pd', 'p95_pd'
]

print("✅ Setup completed. Required 20 feature names loaded:")
print(SELECTED_20_FEATURES)
```

```python
def get_customer_prediction_features(customer_id):
    """
    Accesses the database and extracts exactly 20 features for prediction.
    If customer does not exist in the database, displays:
    'Customer is not in the current database'
    """
    # 1. Verify if customer exists in Customer_Details
    customers = db_manager.fetch_all_customers_db()
    existing_cust_ids = [c['customer_id'] for c in customers]
    
    if int(customer_id) not in existing_cust_ids:
        print(f"Customer is not in the current database")
        return None
    
    # 2. Fetch order transactions from Order_Details table
    orders = db_manager.fetch_order_history_db(int(customer_id))
    
    if not orders:
        print(f"Customer is not in the current database")
        return None
    
    # 3. Load order history into pandas DataFrame
    df_orders = pd.DataFrame(orders)
    
    # Ensure numeric columns are properly typed
    df_orders['product_price'] = df_orders['product_price'].astype(float)
    df_orders['product_discount'] = df_orders['product_discount'].astype(float)
    df_orders['price_after_discount'] = df_orders['price_after_discount'].astype(float)
    
    pp = df_orders['product_price'].values
    pd_disc = df_orders['product_discount'].values
    pad = df_orders['price_after_discount'].values
    
    # 4. Compute exact 20 SHAP features
    feature_dict = {
        'min_pp': float(np.min(pp)),
        'p90_pp': float(np.percentile(pp, 90)),
        'p95_pad': float(np.percentile(pad, 95)),
        'p25_pp': float(np.percentile(pp, 25)),
        'p95_pp': float(np.percentile(pp, 95)),
        'max_pp': float(np.max(pp)),
        'min_pad': float(np.min(pad)),
        'p75_pp': float(np.percentile(pp, 75)),
        'max_pad': float(np.max(pad)),
        'p90_pad': float(np.percentile(pad, 90)),
        'avg_pp': float(np.mean(pp)),
        'p50_pp': float(np.percentile(pp, 50)),
        'p25_pad': float(np.percentile(pad, 25)),
        'p50_pad': float(np.percentile(pad, 50)),
        'avg_pad': float(np.mean(pad)),
        'p75_pad': float(np.percentile(pad, 75)),
        'p90_pd': float(np.percentile(pd_disc, 90)),
        'total_pad': float(np.sum(pad)),
        'max_pd': float(np.max(pd_disc)),
        'p95_pd': float(np.percentile(pd_disc, 95))
    }
    
    # 5. Create 1-row DataFrame with exact feature ordering
    features_df = pd.DataFrame([feature_dict])[SELECTED_20_FEATURES]
    
    print(f"✅ Successfully created 20 features for Customer #{customer_id}:")
    return features_df
```

```python
# Set Customer ID to test (Change customer_id here as needed)
customer_id_input = 1

features_result = get_customer_prediction_features(customer_id_input)

if features_result is not None:
    display(features_result)
    print("\nDataFrame Shape:", features_result.shape)
```

```python
# Test with a Customer ID that does not exist
non_existing_customer_id = 9999

print(f"Testing Customer ID: {non_existing_customer_id}")
features_result_missing = get_customer_prediction_features(non_existing_customer_id)
```

```python
# Optional: Pass extracted 20 features into trained LightGBM model for prediction
if features_result is not None:
    model_path = 'lgbm_multiclass_model.pkl'
    encoder_path = 'label_encoder.pkl'
    
    if os.path.exists(model_path) and os.path.exists(encoder_path):
        with open(model_path, 'rb') as f:
            model = pickle.load(f)
        with open(encoder_path, 'rb') as f:
            encoder = pickle.load(f)
            
        probs = model.predict_proba(features_result)[0]
        predicted_class_encoded = np.argmax(probs)
        predicted_prod_id = encoder.inverse_transform([predicted_class_encoded])[0]
        confidence = probs[predicted_class_encoded] * 100
        
        print(f"🔮 LightGBM Prediction for Customer #{customer_id_input}:")
        print(f"--> Predicted Product ID: #{predicted_prod_id}")
        print(f"--> Confidence Score: {confidence:.2f}%")
    else:
        print("ℹ️ Model files not found for prediction test (lgbm_multiclass_model.pkl / label_encoder.pkl).")
```
