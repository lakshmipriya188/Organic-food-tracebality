```python

c['customer_name'] = (
    c['FirstName'].str.strip() + ' ' +
    c['MiddleInitial'].str.strip() + ' ' +
    c['LastName'].str.strip()
)

# 2. Create email_id
c['email_id'] = (
    c['FirstName'].str[:3].str.lower() +
    c['MiddleInitial'].str[:1].str.lower() +
    c['LastName'].str[-1:].str.lower() +
    c['customer_id'].astype(str) +
    '@gmail.com'
)

# 3. Create password by reversing email_id without @gmail.com
c['password'] = (
    c['email_id']
    .str.replace('@gmail.com', '', regex=False)
    .str[::-1]
)
```
