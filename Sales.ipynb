```python
df.head()
```

```python
# Count how many times each product appears in each category
product_counts = (
    df.groupby(['category_id', 'product_id'])
      .size()
      .reset_index(name='count')
)
product_counts.head(20)
```

```python
#Get top 10 products for each category
top_10_products = (
    product_counts
    .sort_values(['category_id', 'count'], ascending=[True, False])
    .groupby('category_id')
    .head(10)
    .reset_index(drop=True)
)
top_10_products.head(20)
```

```python
# 3. Get only the product IDs that are in the top 10 of their category
top_product = top_10_products[['category_id', 'product_id']]

# 4. Filter the ORIGINAL dataframe
df_top10 = df.merge(
    top_product,
    on=['category_id', 'product_id'],
    how='inner'
)

# 5. View the result
df_top10.shape
```

```python
# Convert order_date
df_top10["order_date"] = pd.to_datetime(
    df_top10["order_date"],
    errors="coerce"
)

```

```python
df_top10 = df_top10[['order_id','customer_id', 'product_id', 'product_count', 'product_price','product_discount', 
                     'price_after_discount', 'order_date','category_id']]
```
